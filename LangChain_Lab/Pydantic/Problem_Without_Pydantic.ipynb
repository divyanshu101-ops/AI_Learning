```python
def add_patient_data(name, age):
    if type(name) == str and type(age) == int:
        print(name)
        print(age)
        print("Data added successfully to the database!")
    else:
        raise ValueError("Invalid data type for name or age. Name shoulb be a string and age should be a integer")
```

```python
add_patient_data("Divyanshu", 25)
```

```python
add_patient_data("Buppy", "twenty five")
```

```python
def add_patient_data1(name, age):
    if type(name) == str and type(age) == int:
        if age >= 0:
            print(name)
            print(age)
            print("Data added successfully to the database!")
        else:
            raise ValueError("Age can not be negative")
    else:
        raise ValueError("Invalid data type for name or age. Name shoulb be a string and age should be a integer")
```

```python
add_patient_data1("Divyanshu", 25)
```

```python
add_patient_data1("Divyanshu", -25)
```
